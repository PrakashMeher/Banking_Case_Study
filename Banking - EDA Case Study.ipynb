```python
bins = [0, 100000, 300000, float('inf')]
labels = ['Low', 'Med', 'High']

df['Income Band'] = pd.cut(df['Estimated Income'], bins=bins, labels=labels, right=False)
```

```python
# Examine the distribution of unique cataegories in categorical columns
categorical_cols = df[["BRId", "GenderId", "Amount of Credit Cards","Nationality","Occupation", "Fee Structure", "Loyalty Classification", "Properties Owned", "Risk Weighting","Income Band"]].columns

for col in categorical_cols:
    print(f"Value Counts for '{col}':")
    display(df[col].value_counts())
```

# Univariate Analysis

```python
for i, predictor in enumerate(df[["BRId", "GenderId", "Amount of Credit Cards","Nationality","Occupation", "Fee Structure", "Loyalty Classification", "Properties Owned", "Risk Weighting","Income Band"]].columns):
    plt.figure(i)
    sns.countplot(data=df, x=predictor)
```

```python
for i, predictor in enumerate(df[["BRId", "GenderId", "Amount of Credit Cards","Nationality","Occupation", "Fee Structure", "Loyalty Classification", "Properties Owned", "Risk Weighting","Income Band"]].columns):
    plt.figure(i)
    sns.countplot(data=df, x=predictor, hue='GenderId')
```

```python
# Histplot of vlue counts for different occupation

for col in categorical_cols:
    if col == "Occupation":
        continue
    plt.figure()
    sns.histplot(df[col])
    plt.title('Histogram of Occupation Count')
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

# Numerical Analysis

```python
numerical_cols = ['Estimated Income', 'Superannuation Savings', 'Credit Card Balance', 'Bank Loans', 'Bank Deposits', 'Checking Accounts', 'Saving Accounts', 'Foreign Currency Account','Business Lending']

# Univariate analysis and visualization
plt.figure(figsize=(15,10))
for i,col in enumerate(numerical_cols):
    plt.subplot(4,3,i+1)
    sns.histplot(df[col], kde=True)
    plt.title(col)
plt.show()   
```

# Heatmaps

```python
numerical_cols = ['Estimated Income', 'Superannuation Savings', 'Credit Card Balance', 'Bank Loans', 'Bank Deposits', 'Checking Accounts', 'Saving Accounts', 'Foreign Currency Account','Business Lending']

correlation_matrix = df[numerical_cols].corr()

plt.figure(figsize=(12,12))
sns.heatmap(correlation_matrix, annot=True, cmap = 'crest', fmt=".2f")
plt.show()
```

# Insights of EDA

1. The strongest positive correlation occur among "Bank Deposits" with "Checking Accounts", "Saving Accounts" and "Foregin Currency Account" indicating that customers who maintain high balances in one account type often hold substational amount/funds across other accounts as well.
